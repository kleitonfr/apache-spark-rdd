# Analisando a Rede Social Youtube

## O objetivo é compreender o comportamento e popularidade de vídeos e dos próprios influenciadores.

```python
!pip install pyspark

from pyspark.sql import SparkSession
from pyspark.sql.types import *
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
videosStatsData = spark.read.csv('youtube_data/videos-stats.csv')
videosStatsData.show(8)
print()
videosStatsData.printSchema()
```

```python
videosStatsData = spark.read.\
option('header', 'true').\
option('inferSchema', 'true').\
csv('youtube_data/videos-stats.csv')

videosStatsData.show()
videosStatsData.printSchema()
```

```python
videosStatsData.write.\
mode('overwrite').\
save('youtube_parquet_data/videos-parquet')

videosParquetData = spark.read.parquet('youtube_parquet_data/videos-parquet')
videosParquetData.show()
```

```python
spark.catalog.listDatabases
```

```python
spark.catalog.setCurrentDatabase('default')
```

```python
spark.catalog.listTables()
```

```python
videosParquetData.write.saveAsTable('tb_videos')
```

```python
spark.catalog.listTables()
```

```python
tab_df = spark.sql('select * from tb_videos')
tab_df.show()
```

```python
commentsData = spark.read.\
option('mode','DROPMALFORMED').\
option('header', 'true').\
option('inferSchema', 'true').\
csv('comments/comments.csv')

commentsData.printSchema()
print()
commentsData.show()
```

```python
commentsData.write.mode('overwrite').save('comments/comments-parquet')
```

```python
commentsParquet = spark.read.parquet('comments/comments-parquet')
commentsParquet.show()
```
